```python
import pandas as pd
import math

df = pd.read_json('./galaxy_1day.json', orient='records')
```

```python
with_bodies = df[ df['bodyCount'] > 0]
```

```python
with_bodies.size
```

```python
populated = with_bodies[with_bodies['population'] > 0 ]
```

```python
unpopulated = with_bodies[with_bodies['population'] <= 0]
```

```python
populated.size
```

```python
def distance(origin):
    return lambda p : math.sqrt((origin.x - p.x)**2 + (origin.y - p.y)**2 + (origin.z - p.z)**2)

def adjecent_systems(origin):
    cp = populated.copy()
    with_dist = cp.assign(dist = math.sqrt((origin['coords']['x'] - cp['coords.x'])**2 + (origin['coords']['y'] - cp['coords.y'])**2 + (origin['coords']['z'] - cp['coords.z'])**2))    
    with_dist = with_dist[with_dist['dist'] <= 15]
    return with_dist
    
```

```python
with_adjecent = unpopulated.assign(adjecent = adjecent_systems)
```

```python
with_adjecent = with_adjecent[with_adjecent['adjecent'].apply(len) > 0]
```
